```python
stud_df1 = spark.read.csv("/Volumes/vivek/default/vkalatest/stud_part4.csv", header=True, inferSchema=True)
stud_df2 = spark.read.csv("/Volumes/vivek/default/vkalatest/stud_part5.csv", header=True, inferSchema=True)
stud_df3 = stud_df1.unionByName(stud_df2,allowMissingColumns=True)
stud_df3.show()

data1 = [(1,"James","Sales",3000),(2,"Michael","Sales",4600),(3,"Robert","Sales",4100),(4,"Maria","Finance",3000)]
data2 = [(1,"James","Sales",3000,2026),(2,"Michael","1111",4600,2026),(3,"Robert","Sales",4100,2026),(4,"Maria","Finance",3000,2026),(5,"Jen","Finance",3000,2026)]
df1 = spark.createDataFrame(data1,['id','name','dept','salary'])
df2 = spark.createDataFrame(data2,['id','name','dept','salary','year'])
df3 = df1.unionByName(df2,allowMissingColumns=True)
df3.show()
df3.filter(df3.salary > 3000).show()
```

```python
schema_str="custid int,fname string,lname string,age int,profession string,error_rec string"
clen_df = spark.read.csv("/Volumes/vivek/default/vkalatest/custsmodified",header=True,schema=schema_str,mode="permissive",columnNameOfCorruptRecord="error_rec")
clean_df = clen_df.filter("error_rec is null AND custid is not null")
clean_df.na.drop(subset=["lname","profession"]).show()
clean_df.na.drop(how="all").show()

#fill
schema_str="custid int,fname string,lname string,age int,profession string,error_rec string"
clen_df = spark.read.csv("/Volumes/vivek/default/vkalatest/custsmodified",header=True,schema=schema_str,mode="permissive",columnNameOfCorruptRecord="error_rec")
not_null_df=clen_df.na.fill("IT",subset=["profession"])
not_null_df.show()  





```
